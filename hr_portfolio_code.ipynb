#importing libraries.

```python

import pandas as pd
import numpy as np
```

#reading the file.

```python
df = pd.read_csv("hr_dirty_portfolio.csv")
```

#checking the file.

```python
display (df)
```

#creating a copy to avoid losing results in the process.

```python
df_original = df.copy()
```

#getting rid of completely empty rows and columns.

```python
df = df.dropna(how = "all")
df = df.dropna(how = "all", axis =1)
```

#getting rid of duplicates.

```python
df = df.drop_duplicates()
```

#replacing empty cells on employee_id column.

```python
df["employee_id"] = df ["employee_id"].replace ("MISSING_ID",pd.NA)
df["employee_id"] = df["employee_id"].fillna (df ["full_name"].str.replace("Employee","EMP", regex = False))
```

#checking result.

```python
df
```

#checking unique values on department column.

```python
df ["department"].unique()
```

#removing extra spaces.

```python
df["department"] = df["department"].str.strip().str.title()
df["department"].unique()
```

#converting missing, unnecessary and negative values to nan, converting valid salary values to numeric values and checking whether any negative salary values remained.

```python

def cleaned_salary(val):
    if pd.isna(val) or val == "CONFIDENTIAL":
        return np.nan
   
   
    num = float(val)
    
    if num > 0:
        return num
    else:
        return np.nan


df["salary"] = df["salary"].apply(cleaned_salary)

df[df["salary"] < 0]
```

#checking the entire file.

```python
df
```

#saving the cleaned file.

```python
df.to_csv('hr_cleaned_portfolio.csv', index=False)
```
